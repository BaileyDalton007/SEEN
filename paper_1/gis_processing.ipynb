```python
import geopandas as gpd
```

```python
shapefile_path = "C:\\Users\\bdsoc\\Downloads\\tl_2023_39_unsd\\tl_2023_39_unsd.shp"
gdf = gpd.read_file(shapefile_path)
```

```python
# Combines all the different LEA columns
def combine_lea_id(row):
    if row['UNSDLEA'] != None:
        return int(row['STATEFP'] + row['UNSDLEA'])
    if row['SDADMLEA'] != None:
        return int(row['STATEFP'] + row['SDADMLEA'])
    if row['SCSDLEA'] != None:
        return int(row['STATEFP'] + row['SCSDLEA'])
    if row ['ELSDLEA'] != None:
        return int(row['STATEFP'] + row['ELSDLEA'])
    
    return None
```

```python
gdf['LEAID'] = gdf.apply(combine_lea_id, axis=1)
```

```python
gdf.to_file("C:\\Users\\bdsoc\\Downloads\\tl_2023_39_unsd\\school_dis.shp", driver='ESRI Shapefile', index=False)
```
